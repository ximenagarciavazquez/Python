```python
#Ejercicio 3 Obtener las raíces e indicar si son reales o complejas
print('Ingrese su ecuación cuadrática de la forma ax^2+bx+c')
a=float(input('ingrese el valor de a:'))
b=float(input('ingrese el valor de b:'))
c=float(input('ingrese el valor de c:'))
discriminante = b**2-4*a*c

es_compleja = (discriminante < 0)
tipos_de_raiz = ['Las raíces son reales']
print(f'\n[tipos_de_raiz{int(es_compleja)}]')

raiz_discriminante=(discriminante)**(1/2)

x1= (-b + raiz_discriminante)/(2*a)
x2= (-b - raiz_discriminante)/(2*a)

x1_final= (x1.real*(1-es_compleja))+(x1*es_compleja)
x2_final= (x2.real*(1-es_compleja))+(x2*es_compleja)

print(f'La solución x1 es: {x1_final}')
print(f'La solución x2 es: {x2_final}')
```
