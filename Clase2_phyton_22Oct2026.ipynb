```python
#Calculadora Simple

valor_del_primer_número = float(input("Escribe el primer número:"))
valor_del_segundo_número = float(input("Escribe el segundo número:"))

suma = valor_del_primer_número + valor_del_segundo_número
resta = valor_del_primer_número - valor_del_segundo_número
multiplicacion = valor_del_primer_número * valor_del_segundo_número

print(f"suma: {suma}")
print(f"resta: {resta}")
print(f"multiplicación: {multiplicacion}")
```

```python
#Conversión de Temperaturas
temperatura_en_grados_celcius = float(input("Escribe la temperatura en grados Celsius: "))

fahrenheit = (temperatura_en_grados_celcius * 1.8) + 32

print(f"Temperatura en grados Fahrenheit: {fahrenheit}")
```

```python
#Cálculo del área de un triángulo
base = float(input("Escribe la base del triángulo: "))
altura = float(input("Escribe la altura del triángulo: "))

area = (base * altura) / 2
print(f"El área del triángulo es: {area}")
```

```python
#Calculadora de Descuentos
precio_del_producto = float(input("Escribe el precio del producto: "))
porcentaje_de_descuento = float(input("Escribe el porcentaje de descuento: "))

descuento = precio_del_producto * (porcentaje_de_descuento / 100)
precio_final = precio_del_producto - descuento

print(f"Precio final: {precio_final}")
```

```python
#Cambio de moneda
dolares = float(input("Dólares: "))
tasa_de_cambio = 0.89

euros = dolares * tasa_de_cambio
print(f"Euros: {euros}")
```

```python
#Calcular el volumen de la esfera V=(4/3)π r^3, r=5
import math

radio = 5.0

volumen = (4/3) * math.pi * (radio * 3)

print(f"El volumen de la esfera con radio {radio} es: {volumen}")
```
